```python
#import basic libraries

import numpy as np
import pandas as pd
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt
```

```python
#Seaborn has inbuilt dataset
sns.get_dataset_names()
```

```python
#load iris datasets

data=sns.load_dataset("iris")
data
```

```python
#Relational plots:are used to understand the releationship between two variables
  # * line plot
  # * scatter plot
  # * replot
```

```python
#line plot
sns.lineplot(data=data,x='sepal_length',y='sepal_width',color="deeppink")

# to use categorical column we use hue which is like a legend

sns.lineplot(data=data,x='sepal_length',y='sepal_width',hue="species",ci=None)
```

```python
#Scatter plot
sns.scatterplot(data=data,x='petal_width',y='petal_length')
plt.show()
```

```python
sns.scatterplot(data=data,x='sepal_length',y='petal_length')
plt.show()
```

```python
sns.scatterplot(data=data,x='sepal_length',y='petal_length',hue="species", palette="rainbow")
plt.show()
```

```python
#relplot
 #relplot by default it is scattered plot
#relplot can be converetd into line plot by giving kind ="line"

sns.relplot(data=data,x='petal_length',y='sepal_width',kind="line",ci=None)
plt.show()
```

#Categorical plots:it is used to analyze one numerical and one categorical column
* count plot
* box plot
* bar plot
* violin plot
* strip plot
* swarm plot
* cat plot





```python
#count plot:basically gives the count of categories
sns.countplot(data=data,x='species', palette="rainbow")
plt.show()
```

```python
#bar plot:
sns.barplot(data=data,x='species',y ='sepal_length')
plt.show()
```

```python
#box plot:it is used to identify outlier
sns.boxplot(data=data,x='sepal_length')
```

```python
sns.boxplot(data=data,x='species',y='sepal_length')
```

```python
#violin plot:it is same as like box plot but itsviolin shape
sns.violinplot(data=data,x="sepal_width")
```

```python
sns.violinplot(data=data,x="species",y='sepal_width')
```

```python
#strip plot:just like scatter plot

sns.stripplot(data=data,x='species' ,y='sepal_width')
```

```python
#swarmplot:overlapping of observations
sns.swarmplot(data=data,x='species', y='sepal_width', hue='species')
```

```python
#cat plot
#using cat plot we can plot box,violin,strip,swarm
#By default it is strip plot
# sns.catplot(data=data,x='species',y='sepal_width',kind='box')
# sns.catplot(data=data,x='species',y='sepal_width',kind='violin')
# sns.catplot(data=data,x='species',y='sepal_width',kind='strip')
sns.catplot(data=data,x='species',y='sepal_width',hue='species',kind='swarm')
```

```python
#Distribution plots:
*histplot
*distplot
 *these two graphs are used to check distribbution of data
```

```python
sns.histplot(data=data,x='petal_width',hue='species',kde=True)
sns.histplot(data=data,x='petal_width',hue='species')
```

```python
#dist plot:
#here we wont use dat=data n kde to conver to curve it automacticlly gives curve
sns.distplot(x=data['sepal_length'])
```

```python
#pair plot
sns.pairplot(data=data,hue='species')
```
