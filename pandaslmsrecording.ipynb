#Data Frames:is a 2Dimentional data structures
* it stores the value in the form of tabular fashion i-e rows n columns
##features of data frames:
* columns are of different data types
* it will have labelled axes(row & columns)
* we can access values rows n columns
* can perform arithematic operations on rows n columns


## syntax is:pd.DataFrame(data,index,columns,dtype) where d n f should b capital



```python
#create a dataframe using nested list

data =[[1,2,3],[4,5,6],[7,8,9]];
A=pd.DataFrame(data)
```

```python
list=[["john",40],["Mary",50],["Arshi",60],["Aalim",40]]
B=pd.DataFrame(list,columns=["Name","Age"],index=[101,102,103,104])
B
```

```python
# creating a dataframe using dict

dict={"Name":["John","Mary","Zara","Ridaa","Arshi"],"Age":[40,50,60,20,30],"Job":["HR","Manager","Analyst","Teacher","Engineer"],"Salary":[30000,40000,50000,20000,60000]}
```

```python
d=pd.DataFrame(dict,index=[101,102,103,104,105])
d
```

```python
#creating a dataframe using nested list

list=[["Apple",20],["Orange",30,120],["Banana",40]]
nestedList =pd.DataFrame(list,columns=["Fruits","No_of_fruits","fruits_per_kilo"])
nestedList # if we dnt have  equal no of values then it will be taken as NAN,null,none values(missing values)
```

```python
#how to sort data
d
d.sort_values("Salary") # by deafault its in ascending value
d.sort_values("Salary",ascending=False,inplace=True)
d
```

```python
#To Sort in Ascending---By default it sorts in ascending order else just give the ascending=True
#To Sort in descending---give ascending =False
# To make permanent change--give inplace=True

#create a copy of data
#copy()-- it helps in creating original copy of data
copyData=d.copy()
copyData

#sort values based on age
copyData.sort_values("Age",inplace=True)
copyData
```

```python
#How to fetch the columns and rows
#loc:loc function is used to fetch columns and rows based on cloumn name or index name.
#iloc:iloc function is used to fetch row and columns based on index position of rows and columns
```

```python
#fetch column name
copyData[["Name"]]
```

```python
#fetch multiple column i-e Age,Salary
copyData[["Age","Salary"]]
```

```python
#we can fetch column name using loc func also
copyData.loc[:,["Name","Age"]]
```

```python
#fetch age using loc func
copyData.loc[:,["Age"]]
```

```python
#fetch name and salary using loc
copyData.loc[:,["Name","Salary"]]
```

```python
#fetch first row names or index name
copyData.loc[[101],:]
```

```python
#fetch the second and third row
copyData.loc[[104,105],:]
```

```python
#fetch first row and last row

copyData.loc[[104,105],:]
```
